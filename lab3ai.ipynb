```python
from collections import deque
import random
import matplotlib.pyplot as plt

# Goal state
GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

MOVES = {
    "Up": -3,
    "Down": 3,
    "Left": -1,
    "Right": 1
}

OPPOSITE = {
    "Up": "Down",
    "Down": "Up",
    "Left": "Right",
    "Right": "Left"
}


# Find legal moves
def actions(state):
    i = state.index(0)
    row, col = divmod(i, 3)
    legal = []

    if row > 0:
        legal.append("Up")
    if row < 2:
        legal.append("Down")
    if col > 0:
        legal.append("Left")
    if col < 2:
        legal.append("Right")

    return legal


# Generate the next state
def result(state, action):
    i = state.index(0)
    j = i + MOVES[action]

    new_state = list(state)
    new_state[i], new_state[j] = new_state[j], new_state[i]

    return tuple(new_state)


# Check solvability
def is_solvable(state):
    tiles = [x for x in state if x != 0]

    inversions = sum(
        1
        for i in range(len(tiles))
        for j in range(i + 1, len(tiles))
        if tiles[i] > tiles[j]
    )

    return inversions % 2 == 0


# Breadth-First Search
def solve_bfs(start):
    if not is_solvable(start):
        return None, 0, 0

    frontier = deque([(start, [])])
    explored = {start}
    expanded = 0
    generated = 1

    while frontier:
        state, path = frontier.popleft()

        if state == GOAL:
            return path, expanded, generated

        expanded += 1

        for action in actions(state):
            child = result(state, action)

            if child not in explored:
                explored.add(child)
                generated += 1
                frontier.append((child, path + [action]))

    return None, expanded, generated


# Generate random solvable states
def random_scramble(number_of_moves, rng):
    state = GOAL
    previous_action = None

    for _ in range(number_of_moves):
        legal = actions(state)

        # Avoid immediately reversing the previous move
        if previous_action is not None:
            legal.remove(OPPOSITE[previous_action])

        action = rng.choice(legal)
        state = result(state, action)
        previous_action = action

    return state


# Use a fixed seed for reproducible results
rng = random.Random(42)

depths = [5, 10, 15, 20, 25]
solution_depths = []
nodes_expanded = []
nodes_generated = []

print("BFS PERFORMANCE COMPARISON")
print("-" * 66)
print(
    f"{'Scramble Depth':<16}"
    f"{'Solution Depth':<16}"
    f"{'Nodes Expanded':<17}"
    f"{'Nodes Generated':<17}"
)
print("-" * 66)

for depth in depths:
    start = random_scramble(depth, rng)

    path, expanded, generated = solve_bfs(start)

    actual_depth = len(path) if path is not None else None

    solution_depths.append(actual_depth)
    nodes_expanded.append(expanded)
    nodes_generated.append(generated)

    print(
        f"{depth:<16}"
        f"{str(actual_depth):<16}"
        f"{expanded:<17}"
        f"{generated:<17}"
    )

    print("Starting state:", start)
    print()


# Plot scramble depth against nodes expanded
plt.figure(figsize=(8, 5))

plt.plot(
    depths,
    nodes_expanded,
    marker="o",
    color="blue",
    linewidth=2
)

plt.title("BFS: Scramble Depth vs Nodes Expanded")
plt.xlabel("Number of Random Moves")
plt.ylabel("Nodes Expanded")
plt.xticks(depths)
plt.grid(True)
plt.tight_layout()

# Save the graph
plt.savefig("bfs_growth.png", dpi=150)

# Display the graph
plt.show()

print("Graph saved as bfs_growth.png")
```

```python
from collections import deque

# Initial and goal states: (missionaries_left,
#                           cannibals_left, boat_position)
INITIAL = (3, 3, 0)
GOAL = (0, 0, 1)

# Number of passengers: (missionaries, cannibals)
ACTIONS = [
    (1, 0),  # One missionary
    (2, 0),  # Two missionaries
    (0, 1),  # One cannibal
    (0, 2),  # Two cannibals
    (1, 1)   # One missionary and one cannibal
]


# Check whether a state is valid and safe
def is_valid(state):
    m_left, c_left, boat = state

    m_right = 3 - m_left
    c_right = 3 - c_left

    # Population limits and boat position
    if not (0 <= m_left <= 3 and 0 <= c_left <= 3):
        return False

    if boat not in (0, 1):
        return False

    # Cannibals cannot outnumber missionaries
    # on either bank when missionaries are present
    if m_left > 0 and c_left > m_left:
        return False

    if m_right > 0 and c_right > m_right:
        return False

    return True


# Goal test
def goal_test(state):
    return state == GOAL


# Return all legal actions from a state
def actions(state):
    m_left, c_left, boat = state
    legal = []

    for m, c in ACTIONS:
        # Boat is on the left bank
        if boat == 0:
            if m <= m_left and c <= c_left:
                new_state = (
                    m_left - m,
                    c_left - c,
                    1
                )

                if is_valid(new_state):
                    legal.append((m, c))

        # Boat is on the right bank
        else:
            m_right = 3 - m_left
            c_right = 3 - c_left

            if m <= m_right and c <= c_right:
                new_state = (
                    m_left + m,
                    c_left + c,
                    0
                )

                if is_valid(new_state):
                    legal.append((m, c))

    return legal


# Transition model
def result(state, action):
    m_left, c_left, boat = state
    m, c = action

    if boat == 0:
        return (m_left - m, c_left - c, 1)

    return (m_left + m, c_left + c, 0)


# Breadth-First Search
def solve_bfs(start):
    if not is_valid(start):
        return None

    frontier = deque([(start, [])])
    explored = {start}

    while frontier:
        state, path = frontier.popleft()

        if goal_test(state):
            return path

        for action in actions(state):
            child = result(state, action)

            if child not in explored:
                explored.add(child)
                frontier.append((child, path + [(action, child)]))

    return None


# Display the solution
def print_solution(start, path):
    if path is None:
        print("No solution found.")
        return

    print("Initial state:", start)
    print_state(start)

    for step, (action, state) in enumerate(path, start=1):
        m, c = action

        print(f"Step {step}: Move {m} missionary/missionaries "
              f"and {c} cannibal(s)")
        print_state(state)

    print("Goal reached!")


# Print the number of people on each bank
def print_state(state):
    m_left, c_left, boat = state
    m_right = 3 - m_left
    c_right = 3 - c_left

    boat_side = "Left" if boat == 0 else "Right"

    print(f"Left bank:  M={m_left}, C={c_left}")
    print(f"Right bank: M={m_right}, C={c_right}")
    print(f"Boat position: {boat_side}")
    print("-" * 35)


# Run the program
path = solve_bfs(INITIAL)

if path is not None:
    print("Solution found!")
    print("Total crossings:", len(path))
    print_solution(INITIAL, path)
else:
    print("No solution exists.")
```
