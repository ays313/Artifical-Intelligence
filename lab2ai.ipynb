```python
import random

def reflex_agent(percept):
    location, status = percept
    if status == "Dirty":
        return "Suck"
    return "Right" if location == "A" else "Left"

def random_agent(percept):
    return random.choice(["Left", "Right", "Suck"])

def make_model_based_agent():
    cleaned = set()
    def agent(percept):
        location, status = percept
        cleaned.add(location)
        if status == "Dirty":
            return "Suck"
        if cleaned == {"A", "B"}:
            return "NoOp"
        return "Right" if location == "A" else "Left"
    return agent

def run(agent, world, location, steps=20, verbose=False):
    world = dict(world)
    score = 0
    for step in range(steps):
        action = agent((location, world[location]))
        if verbose:
            print(f"Step {step:2}: at {location}, sees {world[location]:5} -> {action:5} | score {score}")
        if action == "Suck":
            if world[location] == "Dirty":
                score += 10
            world[location] = "Clean"
        elif action == "Right":
            location = "B"; score -= 1
        elif action == "Left":
            location = "A"; score -= 1
    return score

random.seed(1)
states = [({"A": random.choice(["Clean", "Dirty"]),
            "B": random.choice(["Clean", "Dirty"])},
           random.choice(["A", "B"])) for _ in range(10)]

print("=== TASK 3: reflex vs random ===")
tr = tn = 0
for i, (w, loc) in enumerate(states, 1):
    r = run(reflex_agent, w, loc)
    n = run(random_agent, w, loc)
    tr += r; tn += n
    print(f"State {i:2}: A={w['A']:5} B={w['B']:5} start={loc} | reflex {r:3} | random {n:3}")
print(f"Average reflex: {tr/10}")
print(f"Average random: {tn/10}")

print("\n=== TASK 4: model-based ===")
w, loc = {"A": "Dirty", "B": "Dirty"}, "A"
print("Trace (A Dirty, B Dirty, start A):")
run(make_model_based_agent(), w, loc, steps=6, verbose=True)
tm = tr2 = 0
for w, loc in states:
    tm += run(make_model_based_agent(), w, loc)
    tr2 += run(reflex_agent, w, loc)
print(f"\nAverage reflex:       {tr2/10}")
print(f"Average model-based:  {tm/10}")
```
